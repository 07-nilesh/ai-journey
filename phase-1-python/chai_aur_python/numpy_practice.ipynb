```python
import time
import numpy as np

py_list = list(range(1000000))
np_array = np.arange(1000000)

start = time.time()
result = [x * 2 for x in py_list]
print("List:", time.time() - start)

start = time.time()
result = np_array * 2
print("Array:", time.time() - start)
```
