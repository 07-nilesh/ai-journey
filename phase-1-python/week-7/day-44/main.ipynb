```python
data=pd.read_csv(r"N:\Data\SeoulBikeData.csv",encoding="cp1252")
data.head()
```

```python
data.info()
```

```python
data["Date"]=pd.to_datetime(data["Date"],format="%d/%m/%Y")
```

```python
data["temp_group"]=pd.cut(data["Temperature(°C)"],bins=[-20,20,35,50],labels=["Cold","Moderate","Hot"])
```

```python
data
```

```python
import numpy as np
data["log_bikecount"]=np.log1p(data["Rented Bike Count"])
```

```python
data
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

# Original distribution
plt.subplot(1,2,1)
plt.hist(data["Rented Bike Count"], bins=50, color="skyblue", edgecolor="black")
plt.title("Original Rented Bike Count")

# Log-transformed distribution
plt.subplot(1,2,2)
plt.hist(data["log_bikecount"], bins=50, color="orange", edgecolor="black")
plt.title("Log-Transformed Rented Bike Count")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(data["Date"], data["Rented Bike Count"], label="Original", color="blue")
plt.plot(data["Date"], data["log_bikecount"], label="Log Transformed", color="red")
plt.title("Bike Rentals: Original vs Log Transformed")
plt.xlabel("Date")
plt.ylabel("Count")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=False)
poly_features=poly.fit_transform(data[["Temperature(°C)","Humidity(%)"]])
```

```python
poly_features
```

```python
# Get feature names using the same input names
feature_names = poly.get_feature_names_out(["Temperature(°C)", "Humidity(%)"])

# Convert to DataFrame for clarity
poly_df = pd.DataFrame(poly_features, columns=feature_names)

print(poly_df.head())
```

```python
import matplotlib.pyplot as plt

# Scatter plot: Temperature vs Humidity
plt.figure(figsize=(12,5))

# Original features
plt.subplot(1,2,1)
plt.scatter(data["Temperature(°C)"], data["Humidity(%)"], 
            c="blue", alpha=0.5)
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")

# Interaction feature (Temperature × Humidity)
plt.subplot(1,2,2)
plt.scatter(data["Temperature(°C)"], data["Humidity(%)"], 
            c=data["Temperature(°C)"] * data["Humidity(%)"], cmap="viridis", alpha=0.7)
plt.title("Interaction Feature: Temp × Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.colorbar(label="Temp × Humidity")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

# Temperature squared
plt.subplot(1,2,1)
plt.scatter(data["Temperature(°C)"], data["Humidity(%)"], 
            c=data["Temperature(°C)"]**2, cmap="plasma", alpha=0.7)
plt.title("Temperature² Feature")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.colorbar(label="Temperature²")

# Humidity squared
plt.subplot(1,2,2)
plt.scatter(data["Temperature(°C)"], data["Humidity(%)"], 
            c=data["Humidity(%)"]**2, cmap="cividis", alpha=0.7)
plt.title("Humidity² Feature")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.colorbar(label="Humidity²")

plt.tight_layout()
plt.show()
```
