```python
class BankAccount:
    
    def __init__(self,balance=0):
        self.__balance=balance

    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("deposit amount must be positive")
    def withdraw(self,amount):
        if amount<=self.__balance:
            self.__balance-=amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        
        else:
            print("Insuccient balance")
    def check_balance(self):
        return f"Current balance: {self.__balance}"
rahul=BankAccount(100)
rahul.deposit(10)
rahul.withdraw(90)
rahul.check_balance()

        
```

```python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self):
           
        return self.length*self.width
c=Circle(5)
r=Rectangle(4,6)
c.area()
r.area()
```

```python
class Person:
    def __init__(self,name):
        self.name=name
    def display(self):
        return f"{self.name}"
class Employee(Person):
    def __init__(self, name,empid):
        super().__init__(name)
        self.empid=empid
    def displayinfo(self):
        return f"{self.name},{self.empid}"

class Manager(Person):
    def __init__(self, name,department):
        super().__init__(name)
        self.department=department
    def displaym(self):
        return f"{self.name},{self.department}"
A=Person("nilesh")
A.display()
B=Employee("rahul",420)
B.displayinfo()
C=Manager("rohan","HR")
C.displaym()
```

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    def __str__(self):
        return f"({self.x},{self.y})"
v1=Vector(2,4)
v2=Vector(5,1)
print(v1+v2)
```

```python
class Factory:
    def __init__(self, shape_type, dimension):
        self.shape_type = shape_type
        self.dimension = dimension

    def shape(self):
        if self.shape_type.lower() == "circle":
            return f"Circle with radius {self.dimension}"
        
        elif self.shape_type.lower() == "square":
            return f"Square with side {self.dimension}"
        
        elif self.shape_type.lower() == "rectangle":
            length, width = self.dimension
            return f"Rectangle with length {length} and width {width}"
        
        else:
            return "Unknown shape type"


# Example usage
shape_type = input("Enter shape type (circle/square/rectangle): ")

if shape_type.lower() == "rectangle":
    length = int(input("Enter length: "))
    width = int(input("Enter width: "))
    dimension = (length, width)
else:
    dimension = int(input("Enter dimension: "))

factory = Factory(shape_type, dimension)
print(factory.shape())
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def borrow(self):
        if self.available:
            self.available = False
            return True
        return False

    def return_book(self):
        self.available = True

    def __str__(self):
        status = "Available" if self.available else "Borrowed"
        return f"{self.title} by {self.author} ({status})"


class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.borrow():
            self.borrowed_books.append(book)
            print(f"{self.name} borrowed {book.title}")
        else:
            print(f"{book.title} is not available")

    def return_book(self, book):
        if book in self.borrowed_books:
            book.return_book()
            self.borrowed_books.remove(book)
            print(f"{self.name} returned {book.title}")
        else:
            print(f"{self.name} does not have {book.title}")

    def __str__(self):
        borrowed = ", ".join([b.title for b in self.borrowed_books]) or "None"
        return f"Member: {self.name}, ID: {self.member_id}, Borrowed: {borrowed}"


class Library:
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def list_available_books(self):
        print("Available books:")
        for book in self.books:
            if book.available:
                print(book)


# Example usage
lib = Library()

# Add books
b1 = Book("1984", "George Orwell")
b2 = Book("The Alchemist", "Paulo Coelho")
lib.add_book(b1)
lib.add_book(b2)

# Add members
m1 = Member("Nilesh", 1)
lib.add_member(m1)

# Borrow and return
m1.borrow_book(b1)   # Nilesh borrowed 1984
m1.borrow_book(b2)   # Nilesh borrowed The Alchemist
m1.return_book(b1)   # Nilesh returned 1984

lib.list_available_books()
```

```python
class MenuItem:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def __str__(self):
        return f"{self.name} - Rs {self.price}"
    
class Order:
    def __init__(self):
        self.items=[]
    def add_item(self,item):

        self.items.append(item)
    def remove_item(self,item):
        self.items.remove(item)
    def list_items(self):
        return [str(item) for item in self.items]


class Bill:
    def __init__(self,order):
        self.order=order
    def calculate_total(self):
        return sum(item.price for item in self.order.items)
    def __str__(self):
        items_str = "\n".join(self.order.list_items())
        total = self.calculate_total()
        return f"Bill:\n{items_str}\nTotal: ₹{total}"

    
# Create menu items
pizza = MenuItem("Pizza", 250)
burger = MenuItem("Burger", 150)
coffee = MenuItem("Coffee", 100)

# Create an order and add items
order1 = Order()
order1.add_item(pizza)
order1.add_item(coffee)

# Generate bill
bill1 = Bill(order1)
print(bill1)


```
