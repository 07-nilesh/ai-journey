```python
n=int(input("enter number: "))

if n>1:
    is_prime=True
    for i in range(2,int(n**0.5)+1):
        if n%1==0:
            is_prime= False
            break
    if is_prime:
        print(f"{n} is Prime ✅")
    else:
        print(f"{n} is NOT Prime ❌")
else:
    print(f"{n} is NOT Prime ❌")
    
```

While Loop
The while loop keeps running as long as a condition is True. You use it when you don't know how many times you'll need to repeat.

```python
n=1234
res=0
while n>0:
    res=n%10
    print(res)
    n=n//10
```

```python
n=12345
res=0
while n>0:
    digit=n%10
    res=res*10+digit
    n=n//10
print("Reverse =",res)
```

```python
n=int(input("enter ur number :"))
temp=n
res=0
while temp>0:
    digit=temp%10
    res=res*10+digit
    temp//=10
if res==n:
    print(f"palindrome")
else:
    print("not palindrome")
    
```

```python
ls=["a","b","a","c","b","a"]
d={}
count=0
for ch in ls:
    if ch not in d:
        d[ch]=count
    else:
        count+=1
        d[ch]=count
print(d)

freq={}
for item in ls:
    freq[item]=freq.get(item,0)+1
print(freq)

```
