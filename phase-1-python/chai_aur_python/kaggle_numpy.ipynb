```python
def manual_mul(A,B):
    m,n=A.shape
    n2,p=B.shape

    if n!=n2:
        raise ValueError("error")
    C=np.zeros((m,p),dtype=int)

    for i in range(m):
        for j in range(p):
            for k in range(n):
                C[i][j]+=A[i][k]*B[k][j]
    return C

A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([[7,8],
              [9,10],
              [11,12]])
print(manual_mul(A,B))
print(np.dot(A,B))


```
