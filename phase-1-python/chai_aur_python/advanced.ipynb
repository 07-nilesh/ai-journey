```python
import time
def my_timer(func):
    def wrapper(*args,**kwargs):
        t1=time.time()
        result=func(*args,**kwargs)
        t2=time.time()
        print(f" function ran in {t2-t1} sec ")
        return result
    return wrapper
@my_timer
def display_info(name,age):
    time.sleep(2)
    print(name,age)
display_info("nilesh",22)
```

```python
def authentication(func):
    def wrapper(*args,**kwargs):
        password=int(input("enter password:"))
        if password==1234:
            print("access approved fetching function")
            return func(*args,**kwargs)
            
        else:
            print("access denied wrong password")
            return None
    return wrapper
@authentication
def addition(a,b):
    print(a+b)
addition(2,3)
```

```python
def func_counter(func):
    count=0
    def wrapper():
        nonlocal count
        count+=1
        result=func()
        print(f"Function '{func.__name__}' has been called {count} times")

        return result
    return wrapper

@func_counter
def demo():
    print("hello")

demo()
demo()
```

```python
def casing(func):
    def wrapper(*args):
        result=func(*args)
        res=str(result)
        
        return res.upper()
    return wrapper
@casing
def hello(name):
    return name
hello("nilesh")


```

```python
def memoize(func):
    cache = {}  # dictionary to store results

    def wrapper(*args, **kwargs):
        key = (args, tuple(kwargs.items()))  # make arguments hashable
        if key in cache:
            print(f"Fetching from cache for {func.__name__}{args}{kwargs}")
            return cache[key]
        else:
            result = func(*args, **kwargs)
            cache[key] = result
            print(f"Caching result for {func.__name__}{args}{kwargs}")
            return result
    return wrapper

@memoize
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Example calls
print(fibonacci(10))  # first time → computes
print(fibonacci(10))  # second time → fetches from cache
```

```python
def detail(**kwargs):
    result={}
    for key,value in kwargs.items():
        if isinstance(value,int):
            result[key]=value
    return result

detail(name="nilesh",age=21,course="ai")
```

```python
num=[2, 3, 4, 5, 6, 7, 8, 9]
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return n
prime=list(filter(is_prime,num))
prime
```

```python
from functools import reduce

n=5
def factorial(n):
    if n<1:
        return 1
    return reduce(lambda x,y:x*y,range(1,n+1))
factorial(5)
```
