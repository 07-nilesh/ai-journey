```python
def handle_outliers(data,method="IQR",threshold=3,action="report"):
    data=np.array(data)
    if method =="ZSCORE":
        z_score=(data-np.mean(data))/np.std(data)
        outlier_indices=np.where(np.abs(z_score)>threshold)[0]

    elif method=="IQR":
        q1=np.percentile(data,25)
        q3=np.percentile(data,75)
        iqr=q3-q1
        upper=q3+iqr*1.5
        lower=q1-iqr*1.5
        outlier_indices=np.where((data<lower) | (data>upper))[0]
    outlier_values=data[outlier_indices]
    outlier_values = data[outlier_indices]
    
    if action == "report":
        return outlier_indices, outlier_values
    
    elif action == "remove":
        return data[np.setdiff1d(np.arange(len(data)), outlier_indices)]
    
    elif action == "cap":
        if method == "IQR":
            data = np.clip(data, lower, upper)
        elif method == "Zscore":
            # Cap at mean ± threshold*std
            lower = np.mean(data) - threshold * np.std(data)
            upper = np.mean(data) + threshold * np.std(data)
            data = np.clip(data, lower, upper)
        return data
data = np.array([5, 2, 4.5, 4, 3, 2, 6, 20, 9, 2.5])

# Report outliers using IQR
indices, values = handle_outliers(data, method="IQR", action="report")
print("Outlier Indices:", indices)
print("Outlier Values:", values)

# Remove outliers
cleaned = handle_outliers(data, method="IQR", action="remove")
print("Cleaned Data:", cleaned)

# Cap outliers
capped = handle_outliers(data, method="IQR", action="cap")
print("Capped Data:", capped)

```
