```python
df=pd.read_csv(r"N:\Data\SeoulBikeData.csv",encoding="cp1252")
df.info()
df.shape
```

```python
df["demand_class"]=(df["Rented Bike Count"]>1000).astype(int)
df
```

```python
X=df.drop("demand_class",axis=1)
y=df["demand_class"]

rus=RandomUnderSampler(random_state=42)
X_res,y_res=rus.fit_resample(X,y)
print("original: ",y.value_counts())
print("resampled: ",y_res.value_counts())
```

OverSampling

```python
from imblearn.over_sampling import RandomOverSampler

ros=RandomOverSampler(random_state=42)
X_res,y_res=ros.fit_resample(X,y)
print("Original:", y.value_counts())
print("Resampled:", y_res.value_counts())
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.weekday

```

```python
from imblearn.over_sampling import SMOTE


X = df.drop(["demand_class", "Date"], axis=1)
y = df["demand_class"]

smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)

print("Original:", y.value_counts())
print("Resampled:", y_res.value_counts())
```

```python
from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize Balanced Random Forest
brf = BalancedRandomForestClassifier(n_estimators=100, random_state=42)

# Train
brf.fit(X_train, y_train)

# Predict
y_pred = brf.predict(X_test)

# Evaluate
print(classification_report(y_test, y_pred))
```
