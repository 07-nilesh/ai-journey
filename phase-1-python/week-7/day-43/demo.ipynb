```python
import pandas as pd
import matplotlib.pyplot as plt

# Example Zomato orders
orders = {
    "Customer": ["Amit", "Priya", "Rahul"],
    "OrderTime": ["2020-01-15 12:30", "2020-01-15 19:45", "2020-02-01 09:00"],
    "Amount": [250, 400, 150]
}

df = pd.DataFrame(orders)

# Convert OrderTime to datetime
df["OrderTime"] = pd.to_datetime(df["OrderTime"], format="%Y-%m-%d %H:%M")

# Set datetime index
df.set_index("OrderTime", inplace=True)

# Weekly total sales
weekly_sales = df.resample("W")["Amount"].sum()

# Plot
plt.figure(figsize=(8,4))
weekly_sales.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Weekly Zomato Sales")
plt.xlabel("Week Ending")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

data = {
    "TxnTime": [
        "01-08-2026 10:15", "01-08-2026 12:30", "02-08-2026 18:45",
        "15-08-2026 09:00", "15-08-2026 20:10", "16-08-2026 11:00",
        "05-09-2026 14:20", "10-09-2026 19:30"
    ],
    "Amount": [500, 1200, 300, 700, 1500, 400, 900, 2000],
    "Discount": [50, 100, 20, 70, 200, 30, 90, 300]
}

df = pd.DataFrame(data)
df["TxnTime"] = pd.to_datetime(df["TxnTime"], format="%d-%m-%Y %H:%M")
df.set_index("TxnTime", inplace=True)

# Step 1: Resample to daily frequency (fill missing days with 0)
daily = df.resample("D").sum()

# Step 2: Create lag features (true calendar-based lags)
daily["Amount_lag1"] = daily["Amount"].shift(1)      # yesterday’s amount
daily["Discount_lag2"] = daily["Discount"].shift(2)  # discount two days ago

print(daily.head(20))
```

```python
import pandas as pd

data = {
    "TxnTime": [
        "01-08-2026 10:15", "01-08-2026 12:30", "02-08-2026 18:45",
        "15-08-2026 09:00", "15-08-2026 20:10", "16-08-2026 11:00",
        "05-09-2026 14:20", "10-09-2026 19:30"
    ],
    "Amount": [500, 1200, 300, 700, 1500, 400, 900, 2000],
    "Discount": [50, 100, 20, 70, 200, 30, 90, 300]
}

df = pd.DataFrame(data)
df["TxnTime"] = pd.to_datetime(df["TxnTime"], format="%d-%m-%Y %H:%M")
df.set_index("TxnTime", inplace=True)

# Step 1: Resample daily to fill gaps
daily = df.resample("D").sum()

# Step 2: Rolling mean over last 7 days
daily["Amount_roll7"] = daily["Amount"].rolling("7D").mean()

# Step 3: Shift the rolling mean by 1 day (exclude today)
daily["Amount_roll7_lag1"] = daily["Amount_roll7"].shift(1)

print(daily.head(20))
```
